```python
import json
import os

# Quick test to create a local backup of a notebook format
backup_notebook = {
    "cells": [
        {
            "cell_type": "code",
            "execution_count": None,
            "metadata": {},
            "outputs": [],
            "source": [
                "# Backup Notebook Test\n",
                "print('If you can see this, your notebook file structure is working!')"
            ]
        }
    ],
    "metadata": {},
    "nbformat": 4,
    "nbformat_minor": 0
}

backup_filename = 'backup_notebook.ipynb'
with open(backup_filename, 'w') as f:
    json.dump(backup_notebook, f, indent=2)

print(f"Successfully created local verification file: {os.path.abspath(backup_filename)}")
print("You can find this file in the left-hand folder pane under 'Files' and download it manually.")
```
